### Prime Number

```python
# take input from the user
num = int(input("enter a number:"))
if num <= 1:  # prime numbers are always positive and grater than one
    print(num,"it is not a prime number.")
else:
    # check for factors from 2 to square of number
    for i in range(2,int(num**0.5)+1):
        if num % i == 0:  # if the number is divisible by any number in the range of it's own square root it's not a prime number 
            print(num,"it is not a prime number.")
            break
    else: # if the loop completes without any break, then the given number is prime number
        print(num,"it is a prime number.")
```

### Product Of Random Numbers

```python
import random  # import the random module
num1 = random.randint(1,101)  # providing random integers from 1 to 100
num2 = random.randint(1,101)
# ask the user to perform multiplication function
ans = int(input(f"what is {num1} * {num2}?"))
correct_ans = num1 * num2 # calculate the correct answer
# check if the user answer is correct or not
if ans == correct_ans:
    print("correct! well done.")
else:  # if the user answer is wrong then print the correct answer
    print(f"wrong! the correct ans is {correct_ans}.")
```

### Squares of Even/Odd Numbers

```python
# Creating a function for the even squares from range 100 to 200
def even_square():
    print("Squares of even numbers between 100 and 200:")
    for i in range(100, 201):
        if i % 2 == 0:  # cheeking for even numbers
            print(f'{i}^2 = {i**2}')
# Creating a function for the odd squares from range 100 to 200
def odd_square():
    print("Squares of odd numbers between 100 and 200:")
    for i in range(100, 201):
        if i % 2 != 0:  # cheeking for odd numbers
            print(f'{i}^2 = {i**2}')
# Creating a fucnction for the type of numbers even or odd
def square_type():
    # takeing a input from the user
    user_choice = input("Enter Even or Odd: ").lower()

    if user_choice in ['even', 'even numbers']:# verifing and gives even squares
        even_square()
    elif user_choice in ['odd', 'odd numbers']:# verifing and gives odd squares
        odd_square()
    else:
        print("Invalid choice! Program ended.")

# Call the main function
square_type()
```

### Word Counter

```python
input_text = "Python is a powerful programming language. Python makes coding simple and fun."
for ch in ".,":
    input_text=input_text.replace(ch, "")

input_text=input_text.lower()
print(input_text)
words = input_text.split()
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

### Check for Polyndrome

```python
def clean_text(text):
    # Keep only alphanumeric characters and convert to lowercase
    return ''.join(char.lower() for char in text if char.isalnum())

def is_palindrome(s):
    # Base case: if the string is empty or has one character
    if len(s) <= 1:
        return True
    # Recursive case: compare first and last characters
    if s[0] == s[-1]:
        return is_palindrome(s[1:-1])
    else:
        return False

# Take input from user
user_input = input("Enter a word or phrase: ")

# Clean the input text
cleaned_input = clean_text(user_input)

# Call the recursive function and print the result
if is_palindrome(cleaned_input):
    print("It's a palindrome!")
else:
    print("It's not a palindrome.")
```
