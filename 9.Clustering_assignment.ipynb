### Implementing Clustering Algorithms:

```python
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
from sklearn.metrics import silhouette_score
```

```python
# Elbow curve 

intertia = []
k_values = range(1,10)

for k in k_values:
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(data)
    intertia.append(kmeans.inertia_)
```

```python
plt.plot(k_values,intertia,'bo-')
plt.xlabel('number of clusters (k)')
plt.ylabel('inertia')
plt.title('elbow method for optimal k')
plt.show()
```

```python
# applying KMeans

kmeans = KMeans(n_clusters=3,random_state=42)
kmeans_labels = kmeans.fit_predict(data)

print('silhouette score (k-means):',silhouette_score(data,kmeans_labels))
```

### Hierarchical Clustering 

```python
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
# dendrogram

linkage_methods = ['single', 'complete', 'average', 'ward']

for method in linkage_methods:
    linked = linkage(data, method=method)

    plt.figure(figsize=(6,4))
    dendrogram(linked)
    plt.title(f"Dendrogram ({method} linkage)")
    plt.show()
```


### Hierarchical Clustering Analysis

- **Single linkage**: suffers from chaining effect
- **Complete linkage**: produces compact clusters
- **Average linkage**: balanced approach
- **Ward linkage**: minimizes variance and gives best clusters

**Best method:** Ward linkage produced the most meaningful clusters.


```python
from sklearn.cluster import AgglomerativeClustering

X = data.values 

for method in ['single', 'complete', 'average', 'ward']:
    hc = AgglomerativeClustering(n_clusters=3, linkage=method)
    labels = hc.fit_predict(X)

    plt.scatter(X[:,0], X[:,1], c=labels, cmap='rainbow')
    plt.title(f"Hierarchical Clustering - {method}")
    plt.show()
```

```python
# applying hierarchical clustering

hierarchical = AgglomerativeClustering(n_clusters=3,linkage='average'
)
hier_labels = hierarchical.fit_predict(data)

print('Silhouette score (Hierarchical):',silhouette_score(data,hier_labels))
```

### DBSCAN 

```python
# epsilon

dbscan = DBSCAN(eps=0.5,min_samples=5)
dbscan_labels = dbscan.fit_predict(data)
```

```python
eps_values = [0.3, 0.5, 0.7, 1.0]
min_samples_values = [3, 5, 10]

for eps in eps_values:
    for min_samples in min_samples_values:
        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan.fit_predict(X)

        plt.scatter(X[:,0], X[:,1], c=labels, cmap='rainbow')
        plt.title(f"DBSCAN eps={eps}, min_samples={min_samples}")
        plt.show()
```


### DBSCAN Analysis

From the experiments:

- When `eps` is too small, many points are classified as noise.
- Increasing `eps` reduces noise but may merge clusters.
- Higher `min_samples` increases robustness but also increases noise.

**Best parameters (based on visualization):**
- eps = 0.5 (example)
- min_samples = 5 (example)

These values produce well-separated clusters with minimal noise.


```python
# applying DBSCAN

n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
print('DBSCAN cluster:',n_clusters)

if n_clusters > 1:
    score = silhouette_score(data,dbscan_labels)
    print('silhouette score (DBSCAN):',score)
else:
    print('Silhouette score not applicables (only one cluster found)')
```
